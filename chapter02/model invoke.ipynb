# 关于模型调用的

## invoke

```python
#调用本地大模型
#invoke 是阻塞式的 一次性返回全部（可以传文本 字典 消息对象列表）
#stream是流式 一个一个字
#batch 批量处理
import os
from langchain_ollama import ChatOllama
#初始化大模型
llm=ChatOllama(
    model="qwen3.8:27b-mlx",
    temperature=0.6,#控制大模型输出的随机性和创造性。取值范围0～1之间 越大越创造
)
```

```python
from email import message

## 文本输入
message=[
    {"role":"system","content":"你是一个专业的数学老师"},
    {
        "role":"user","content":"帮我解释一下什么是斐波那契数列" }
]
response = llm.invoke(message)
print(response.content)
```

```python
##加上记忆
from email import message

## 文本输入
message1=[
    {"role":"system","content":"你是一个 ai 助手"},
    {"role":"user","content":"你好 我叫小明" }
]
#第一次对话
response1 = llm.invoke(message1)
print(f"ai回复 1：{response1.content}")
#第二次对话

#添加记忆
message1.append({"role":"assistant","content":response1.content})
message1.append(
    { "role":"user","content":"我叫什么名字"}
)
response2 = llm.invoke(message1)
print(f"ai回复 2：{response2}")
```
